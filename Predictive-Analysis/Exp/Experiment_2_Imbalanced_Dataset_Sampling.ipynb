# Experiment 2: Handling Imbalanced Datasets (Random Undersampling & Oversampling)
**Aim:** To identify class imbalance, apply undersampling and oversampling techniques, and compare class distributions before and after balancing.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler
```

### A) Data Inspection & Stratified Train-Test Split

```python
# Load dataset (update path if needed, e.g., 'Datasets/imbalanced_dataset.csv')
df = pd.read_csv("../Datasets/imbalanced_dataset.csv")

print("Shape:", df.shape)
print(df["Class"].value_counts().sort_index())

df["Class"].value_counts().sort_index().plot(
    kind="bar", figsize=(4, 3), title="Before Balancing"
)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

X = df.drop(columns="Class")
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

### B) Random Undersampling

```python
rus = RandomUnderSampler(random_state=42)
X_under, y_under = rus.fit_resample(X_train, y_train)

print("Class Distribution After Undersampling:")
print(y_under.value_counts().sort_index())

y_under.value_counts().sort_index().plot(
    kind="bar", figsize=(4, 3), title="After Undersampling"
)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

### C) Random Oversampling

```python
ros = RandomOverSampler(random_state=42)
X_over, y_over = ros.fit_resample(X_train, y_train)

print("Class Distribution After Oversampling:")
print(y_over.value_counts().sort_index())

y_over.value_counts().sort_index().plot(
    kind="bar", figsize=(4, 3), title="After Oversampling"
)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
