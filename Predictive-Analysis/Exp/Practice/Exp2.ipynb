```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler
```

```python
df = pd.read_csv("../Datasets/imbalanced_dataset.csv")
df.sample(5)
```

```python
df.shape
```

```python
df["Class"].value_counts().plot(kind="bar", title="Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
X = df.drop(columns="Class")
y = df["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
rus = RandomUnderSampler(random_state=42)
X_under, y_under = rus.fit_resample(X_train, y_train)

print("After Under-sampling: ", y_under.value_counts().sort_index())

y_under.value_counts().sort_index().plot(
    kind="bar", figsize=(4, 3), title="After Undersampling"
)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
ros = RandomOverSampler(random_state=42)
X_over, y_over = ros.fit_resample(X_train, y_train)

print("After Over-sampling: ", y_over.value_counts().sort_index())

y_over.value_counts().sort_index().plot(
    kind="bar", figsize=(4, 3), title="After Oversampling"
)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
