# Experiment 4: Student Pass/Fail Prediction using Logistic Regression
**Aim:** To develop a Logistic Regression model to predict student pass/fail outcomes and evaluate its performance using Accuracy, Precision, Recall, F1-score, and a confusion matrix.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, confusion_matrix,
    precision_score, recall_score, f1_score
)
```

### A) Dataset Loading and Preprocessing

```python
# Load dataset (update path if needed, e.g., '../Datasets/student_pass_fail_final.csv')
df = pd.read_csv("student_pass_fail_final.csv")

print("Original Shape:", df.shape)
print("Missing Values:\n", df.isnull().sum())
print("Duplicate Rows:", df.duplicated().sum())

df.fillna(df.mean(numeric_only=True), inplace=True)
df.drop_duplicates(inplace=True)

print("Cleaned Shape:", df.shape)
print("Remaining Missing Values:", df.isnull().sum().sum())
```

### B) Model Training

```python
X = df.drop(columns=["Pass"])
y = df["Pass"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.8, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

print("Model trained successfully.")
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
```

### C) Prediction and Evaluation

```python
y_pred = model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="weighted", zero_division=0))
print("Recall:", recall_score(y_test, y_pred, average="weighted", zero_division=0))
print("F1-Score:", f1_score(y_test, y_pred, average="weighted", zero_division=0))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(4, 3))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=["Fail", "Pass"],
    yticklabels=["Fail", "Pass"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

### D) Sample Student Pass/Fail Prediction

```python
student = pd.DataFrame([[8, 8, 2, 1, 67, 72]], columns=X.columns)
prediction = model.predict(scaler.transform(student))[0]

print("Student Details:")
print("Study Hours: 8")
print("Sleeping Hours: 8")
print("Social Media Usage: 2")
print("Playing Hours: 1")
print("Previous Score: 67")
print("Attendance: 72")
print("Final Prediction:", "Pass" if prediction == 1 else "Fail")
```
