```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder

data = {
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue', 'Green', 'Red', 'Green'],
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Large', 'Small', 'Large', 'Small'],
    'Target': [1, 0, 1, 1, 0, 0, 1, 1]
}

df = pd.DataFrame(data)

df
```

```python
le = LabelEncoder()

df['Color_LabelEncoded'] = le.fit_transform(df['Color'])

df
```

```python
ord_enc = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])

df['Size_OrdinalEncoded'] = ord_enc.fit_transform(df[['Size']])

df
```

```python
data = pd.DataFrame({
    'Education': ['High School', 'Bachelor', 'Master', 'PhD', 'Bachelor']
})

data

education_mapping = {
    'High School': 0,
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3
}

data['Education_OrdinalEncoded'] = data['Education'].map(education_mapping)

print(data)
```

```python
ohe = OneHotEncoder()

ohe_df = pd.DataFrame()
ohe.fit_transform(df[['Color']]),
columns=ohe.get_feature_names_out(['Color'])


df = pd.concat([df, ohe_df], axis=1)

df
```

```python
import pandas as pd

data = pd.DataFrame({
    'Vehicle': ['Car', 'Bike', 'Truck', 'Bike', 'Car']
})

data
```

```python
data_one_hot = pd.get_dummies(
    data,
    columns=['Vehicle'],
    prefix='Vehicle',
    drop_first=True
)

# Display the result
print(data_one_hot)
```
