```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.001, iterations=1000):
        self.learning_rate = learning_rate
        self.iterations = iterations
        self.weight = None
        self.bias = None
        
    def _sigmoid(self, z):
        """
        Helper method to compute the sigmoid of z.
        The underscore indicates it's meant for internal use by the class.
        """
        return 1 / (1 + np.exp(-z))
        
    def fit(self, X, y):
        # 1. Initialization
        n_samples, n_features = X.shape
        self.weight = np.zeros(n_features)
        self.bias = 0
        
        # 2. Gradient Descent Loop
        for _ in range(self.iterations):
            # TODO: Calculate the linear equation (z)
            linear_model = np.dot(X, self.weight) + self.bias
            
            # TODO: Apply the sigmoid function to get predictions (y_predicted)
            y_predicted = self._sigmoid(linear_model)
            
            # (We will add the gradients and update steps here in the next task!)
            dw = np.dot(X.T, (y_predicted - y)) / n_samples
            db = np.sum(y_predicted - y) / n_samples
            self.weight = self.weight - (self.learning_rate * dw)
            self.bias = self.bias - (self.learning_rate * db)

        
    def predict(self, X):
        # 1. Run the forward pass exactly like in fit
        linear_model = np.dot(X, self.weight) + self.bias
        y_predicted = self._sigmoid(linear_model)
        
        # 2. Convert probabilities to class labels using a list comprehension
        # If the probability is > 0.5, classify as 1, else 0
        y_predicted_cls = [1 if i > 0.5 else 0 for i in y_predicted]
        
        return np.array(y_predicted_cls)
```

```python
from sklearn.model_selection import train_test_split
from sklearn import datasets

# Load a real dataset (Breast Cancer dataset - binary classification)
dataset = datasets.load_breast_cancer()
X, y = dataset.data, dataset.target

print("Features shape (Rows, Columns):", X.shape)
print("Target shape:", y.shape)
```

```python
# Split into training and testing sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1234
)
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
# Instantiate our custom model
model = LogisticRegression(learning_rate=0.0001, iterations=1000)

# Train the model
print("Training model...")
model.fit(X_train, y_train)
print("Model training complete!")
```

```python
# Predict on the test set
predictions = model.predict(X_test)

# Calculate accuracy
def accuracy(y_true, y_pred):
    accuracy = np.sum(y_true == y_pred) / len(y_true)
    return accuracy

acc = accuracy(y_test, predictions)
print(f"Our Custom Logistic Regression Accuracy: {acc * 100:.2f}%")
```
