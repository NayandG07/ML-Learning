### Simple Linear Regression

```python
import numpy as np

class SimpleLinearRegression:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weight = None
        self.bias = None

    def fit(self, X, y):
        """
        This is where the model learns the relationship between X and y.
        """
        # Step 1: Initialize parameters (weight and bias)
        # TODO: Set self.weight and self.bias to starting values (e.g., 0)
        # Step 2: Implement the training loop for self.n_iterations
        # TODO: Inside the loop, calculate predictions
        # TODO: Calculate the gradients (derivatives)
        # TODO: Update self.weight and self.bias
        
        self.weight = 0
        self.bias = 0
        for i in range(self.n_iterations):
            y_predicted = self.predict(X)
            error = y_predicted - y 
            dw = np.sum(2 * X * error)/len(X)
            db = np.sum(2 * error) / len(X)
            self.weight = self.weight - (self.learning_rate * dw)
            self.bias = self.bias - (self.learning_rate * db)

    def predict(self, X):
        # Given new X values, predict the corresponding y values.
        # 1. Access the weight from the class using 'self'
        # 2. Access the bias from the class using 'self'
        # 3. Multiply the input 'X' by the weight
        # 4. Add the bias to that result
        # 5. Return the final calculation
        w = self.weight
        b = self.bias
        y_predicted = w*X + b

        return y_predicted
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])
```

```python
model = SimpleLinearRegression(learning_rate=0.01, n_iterations=1000)
```

```python
model.fit(x, y)
```

```python
print(f"Learned Weight: {model.weight},\nLearned Bias: {model.bias}")
```
