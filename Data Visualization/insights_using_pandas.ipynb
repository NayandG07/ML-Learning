```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)

df.head()
```

```python
df.head()
```

```python
# 1) Quick overview
# df.head()
df.info()
df.describe()
```

```python
df.describe(include="all")
```

```python
# 2) Missing data
df.isna().sum()
```

```python
df.isna().mean().sort_values(ascending=False)
```

```python
# 3) Value distributions (category)
df["origin"].value_counts()
```

```python
# Table of total cars by origin
df["origin"].value_counts().reset_index(name="total_cars").rename(columns={"index": "origin"})
```

```python
df["cylinders"].value_counts()
```

```python
# 4) Group comparisons
df.groupby("origin")["mpg"].mean().sort_values(ascending=False)
```

```python
df.groupby("cylinders").agg(count=("mpg", "size"), avg_mpg=("mpg", "mean"))
```

```python
# 5) Correlations
df.corr(numeric_only=True)
```

```python
# 6) Top/bottom records
df.nlargest(5, "mpg")[["name", "mpg", "horsepower", "weight"]]
```

```python
df.nsmallest(5, "mpg")[["name", "mpg", "horsepower", "weight"]]
```

```python
# 7) Pivot table
pd.pivot_table(df, index="origin", columns="cylinders", values="mpg", aggfunc="mean")
```

```python
# 8) “Time series” style trend by model year
df.groupby("model_year")["mpg"].mean()
```
