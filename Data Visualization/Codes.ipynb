```python
import numpy as np
# 1. Creation of 1D and 2D arrays
sales_1d = np.array([5000, 7000, 8000, 6500, 9000])
sales_2d = np.array([
    [5000, 7000, 8000],
    [6500, 7200, 8100],
    [4000, 6000, 7500]
])
print("1D Array - Monthly Sales:")
print(sales_1d)
print("2D Array - Sales Data:")
print(sales_2d)
```

```python
# 2. Indexing and slicing
print("Indexing:")
print("First month sales:", sales_1d[0])
print("Profit of third product:", sales_2d[2, 2])
print("Slicing:")
print("Sales from month 2 to month 4:", sales_1d[1:4])
```

```python
# 3. Mathematical and statistical operations
print("Mathematical and Statistical Operations:")
print("Total Sales:", np.sum(sales_1d))
print("Average Sales:", np.mean(sales_1d))
print("Maximum Sales:", np.max(sales_2d))
print("Minimum Sales:", np.min(sales_2d))
```

```python
# 4. Element-wise arithmetic
print("Sales after Rs500 Bonus:\n1D Array:\n", sales_1d + 500, "\n2D Array:\n", sales_2d + 500)
```

```python
import matplotlib.pyplot as plt
# Sales Data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
monthly_sales = [5000, 7000, 8000, 6500, 9000]
products = ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard']
product_sales = [50000, 30000, 20000, 15000, 10000]
```

```python
# 1. Line Graph
plt.figure(figsize=(5,3))
plt.plot(months, monthly_sales, marker='o', label="Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.legend()
plt.show()
```

```python
# 2. Bar Chart
plt.figure(figsize=(5,3))
plt.bar(months, monthly_sales, color="skyblue")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Product-wise Sales")
plt.show()
```

```python
plt.figure(figsize=(3,3))
# 3. Pie Chart
plt.pie(product_sales, labels=products, autopct="%1.1f%%")
plt.title("Product-wise Distribution")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Load dataset
df = pd.read_csv("smartmart_sales.csv")
df
```

```python
# 1. Histogram with KDE
plt.figure(figsize=(3,3))
sns.histplot(df["Sales"], kde=True)
plt.title("Sales Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,3))
# 2. Scatterplot with Hue
sns.scatterplot(x="Sales", y="Profit", hue="Category", data=df)
plt.title("Sales vs Profit by Category")
plt.show()
```

```python
plt.figure(figsize=(8,2))
# 3. Heatmap for Correlation
correlation = df.corr(numeric_only=True)
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,3))
# 4. Boxplot for Distribution Analysis
sns.boxplot(x="Category", y="Sales", data=df)
plt.title("Sales Distribution by Category")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset("titanic")
```

```python
titanic.head(5)
```

```python
titanic.info()
```

```python
titanic.describe()
```

```python
# Insights Visualization
plt.figure(figsize=(4,3))
survival_count = titanic["survived"].value_counts()
plt.pie(survival_count, labels=["Not Survived", "Survived"], autopct="%1.1f%%")
plt.title("Survival Rate")
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x="sex", hue="survived", data=titanic)
plt.title("Gender-wise Survival Count")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
print("Female passengers had a significantly higher survival rate compared to male passengers.")
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x="class", hue="survived", data=titanic)
plt.title("Survival Count by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
print("Most passengers in the 1st class survived, while most in the 3rd class did not survive.")
```

```python
# Age distribution
plt.figure(figsize=(4,3))
sns.histplot(titanic["age"].dropna(), bins=30, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
print("The age distribution shows that most passengers were between 20 and 40 years old.")
```

```python
# Fare vs Survival
plt.figure(figsize=(4,3))
sns.boxplot(x="survived", y="fare", data=titanic)
plt.title("Fare vs Survival")
plt.xlabel("Survived")
plt.ylabel("Fare")
plt.show()
print("Passengers who survived tended to pay a higher fare compared to those who did not survive.")
```
