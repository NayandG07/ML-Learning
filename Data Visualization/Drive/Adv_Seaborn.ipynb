```python
# ============================================================
# DATA VISUALIZATION USING SEABORN LIBRARY
# ============================================================

# Seaborn is a Python data visualization library.
# It is built on top of Matplotlib.
# It is mainly used for statistical plots.
# It gives beautiful and meaningful graphs with less code.

# Why do we use Seaborn?
# 1. It makes graphs attractive.
# 2. It works very well with Pandas DataFrame.
# 3. It has many built-in statistical plots.
# 4. It is useful for EDA: Exploratory Data Analysis.
# 5. It helps us find patterns, trends, relationships and outliers.
```

```python
# ============================================================
# IMPORTING REQUIRED LIBRARIES
# ============================================================

# pandas is used for working with data tables.
import pandas as pd

# numpy is used for numerical operations.
import numpy as np

# matplotlib is the base plotting library.
import matplotlib.pyplot as plt

# seaborn is used for advanced and attractive visualization.
import seaborn as sns

# This line helps us display graphs inside Jupyter Notebook.
%matplotlib inline
```

```python
# ============================================================
# LOADING BUILT-IN DATASETS FROM SEABORN
# ============================================================

# Seaborn provides some built-in datasets for practice.
# We will use these datasets to understand different types of plots.

tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
penguins = sns.load_dataset("penguins")
titanic = sns.load_dataset("titanic")

# Display first 5 rows of tips dataset.
tips.head()
```

```python
# ============================================================
# UNDERSTANDING THE TIPS DATASET
# ============================================================

# The tips dataset contains restaurant bill information.

# total_bill : Total bill amount paid by customer
# tip        : Tip amount given by customer
# sex        : Gender of customer
# smoker     : Whether customer is smoker or non-smoker
# day        : Day of visit
# time       : Lunch or Dinner
# size       : Number of people at the table

tips.info()
```

```python
# Display first 10 rows of tips dataset.

tips.head(10)
```

```python
# ============================================================
# BASIC STATISTICAL SUMMARY
# ============================================================

# describe() gives count, mean, standard deviation,
# minimum, maximum and quartile values for numerical columns.

tips.describe()
```

```python
# ============================================================
# RELATIONAL PLOTS
# ============================================================

# Relational plots are used to show relationship between two variables.
# Mostly, both variables are numerical.

# Example:
# Does total bill increase when tip increases?
# Does passenger count increase over years?
# Does age affect salary?

# Common relational plots in Seaborn:
# 1. Scatter Plot
# 2. Line Plot
# 3. lmplot
```

```python
# ============================================================
# SCATTER PLOT
# ============================================================

# Scatter plot is used to show relationship between two numerical variables.
# Each point represents one record.

# Here:
# x-axis = total_bill
# y-axis = tip

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Scatter Plot: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()

# Observation:
# As total_bill increases, tip also generally increases.
```

```python
# ============================================================
# SCATTER PLOT WITH HUE
# ============================================================

# hue is used to divide data using different colors.
# Here, smoker and non-smoker customers are shown in different colors.

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="smoker")

plt.title("Scatter Plot with Hue: Smoker Wise")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# ============================================================
# SCATTER PLOT WITH HUE, STYLE AND SIZE
# ============================================================

# hue   : Changes color according to category.
# style : Changes marker style according to category.
# size  : Changes size of points according to value.

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    size="size"
)

plt.title("Scatter Plot with Hue, Style and Size")
plt.show()
```

```python
# ============================================================
# LINE PLOT
# ============================================================

# Line plot is used to show trend over time.
# It is mostly used when x-axis has time-related values.

# First, check flights dataset.
flights.head()
```

```python
# Here we show passenger trend year-wise.

sns.lineplot(data=flights, x="year", y="passengers")

plt.title("Line Plot: Passengers Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.show()

# Observation:
# Number of passengers increased over years.
```

```python
# ============================================================
# LINE PLOT WITH HUE
# ============================================================

# hue='month' creates a separate line for each month.

sns.lineplot(data=flights, x="year", y="passengers", hue="month")

plt.title("Line Plot: Monthly Passenger Trend")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()
```

```python
# ============================================================
# LMPLOT
# ============================================================

# lmplot means linear model plot.
# It creates scatter plot with regression line.
# Regression line shows the general trend.

sns.lmplot(data=tips, x="total_bill", y="tip")

plt.title("Lmplot: Total Bill vs Tip")
plt.show()
```

```python
# ============================================================
# CATEGORICAL PLOTS
# ============================================================

# Categorical plots are used when one variable is categorical.
# A categorical variable contains groups/classes.

# Examples of categorical variables:
# day, sex, smoker, time, species

# Common categorical plots:
# 1. Stripplot
# 2. Swarmplot
# 3. Barplot
# 4. Countplot
# 5. Boxplot
# 6. Violinplot
```

```python
# ============================================================
# STRIPPLOT
# ============================================================

# Stripplot shows individual data points for each category.
# It is useful when we want to see each value separately.

sns.stripplot(data=tips, x="day", y="total_bill")

plt.title("Stripplot: Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# ============================================================
# STRIPPLOT WITH JITTER
# ============================================================

# Problem:
# Some points may overlap each other.

# Solution:
# jitter=True spreads points slightly so values become visible.

sns.stripplot(data=tips, x="day", y="total_bill", jitter=True)

plt.title("Stripplot with Jitter")
plt.show()
```

```python
# ============================================================
# STRIPPLOT WITH HUE
# ============================================================

# hue divides the points using color.

sns.stripplot(data=tips, x="day", y="total_bill", hue="sex", jitter=True)

plt.title("Stripplot: Day Wise Bill by Gender")
plt.show()
```

```python
# ============================================================
# SWARMPLOT
# ============================================================

# Swarmplot is similar to stripplot.
# Difference:
# Swarmplot adjusts points so that they do not overlap.

# It is useful for small datasets.
# For large datasets, swarmplot can become slow.

sns.swarmplot(data=tips, x="day", y="total_bill")

plt.title("Swarmplot: Total Bill by Day")
plt.show()
```

```python
# ============================================================
# SWARMPLOT WITH HUE
# ============================================================

sns.swarmplot(data=tips, x="day", y="total_bill", hue="sex")

plt.title("Swarmplot: Total Bill by Day and Gender")
plt.show()
```

```python
# ============================================================
# BARPLOT
# ============================================================

# Barplot is used to compare numerical values across categories.
# By default, Seaborn barplot shows the mean value.

# Here it shows average total bill for each day.

sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Barplot: Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()
```

```python
# ============================================================
# BARPLOT WITH HUE
# ============================================================

# Here we compare average total bill day-wise and gender-wise.

sns.barplot(data=tips, x="day", y="total_bill", hue="sex")

plt.title("Barplot: Average Bill by Day and Gender")
plt.show()
```

```python
# ============================================================
# COUNTPLOT
# ============================================================

# Countplot counts the number of records in each category.
# It is used when we want frequency/count.

# Here it counts number of customers for each day.

sns.countplot(data=tips, x="day")

plt.title("Countplot: Number of Customers by Day")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
# ============================================================
# COUNTPLOT WITH HUE
# ============================================================

# This counts customers day-wise and divides them by gender.

sns.countplot(data=tips, x="day", hue="sex")

plt.title("Countplot: Customers by Day and Gender")
plt.show()
```

```python
# ============================================================
# BOXPLOT
# ============================================================

# Boxplot shows distribution of numerical data.
# It is useful to identify spread and outliers.

# Boxplot shows:
# 1. Minimum value
# 2. First quartile Q1
# 3. Median
# 4. Third quartile Q3
# 5. Maximum value
# 6. Outliers

sns.boxplot(data=tips, x="day", y="total_bill")

plt.title("Boxplot: Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# ============================================================
# BOXPLOT WITH HUE
# ============================================================

sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")

plt.title("Boxplot: Total Bill by Day and Gender")
plt.show()
```

```python
# ============================================================
# VIOLINPLOT
# ============================================================

# Violinplot is a combination of boxplot and density plot.
# It shows distribution shape of data.

# Wider area means more values are present there.
# Narrower area means fewer values are present there.

sns.violinplot(data=tips, x="day", y="total_bill")

plt.title("Violinplot: Total Bill by Day")
plt.show()
```

```python
# ============================================================
# VIOLINPLOT WITH HUE
# ============================================================

sns.violinplot(data=tips, x="day", y="total_bill", hue="sex")

plt.title("Violinplot: Total Bill by Day and Gender")
plt.show()
```

```python
# ============================================================
# VIOLINPLOT WITH SPLIT
# ============================================================

# split=True can be used when hue has only two categories.

sns.violinplot(data=tips, x="day", y="total_bill", hue="sex", split=True)

plt.title("Split Violinplot")
plt.show()
```

```python
# ============================================================
# DISTRIBUTION PLOTS
# ============================================================

# Distribution plots are used to understand how values are spread.

# They answer questions like:
# Are most values small or large?
# Is the data normally distributed?
# Is the data skewed?
# Are there outliers?

# Common distribution plots:
# 1. Histplot
# 2. Displot
# 3. Jointplot
# 4. Pairplot
# 5. Rugplot
```

```python
# ============================================================
# HISTPLOT
# ============================================================

# Histplot shows distribution of a numerical variable using bars.
# Bars divide the data into intervals called bins.

sns.histplot(data=tips, x="total_bill")

plt.title("Histplot: Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# ============================================================
# HISTPLOT WITH KDE
# ============================================================

# kde=True adds a smooth curve over histogram.
# KDE means Kernel Density Estimate.

sns.histplot(data=tips, x="total_bill", kde=True)

plt.title("Histplot with KDE")
plt.show()
```

```python
# ============================================================
# DISTPLOT / DISPLOT
# ============================================================

# Earlier, Seaborn used sns.distplot().
# But now distplot is deprecated.
# Deprecated means it is old and not recommended.

# Instead of distplot, we use:
# sns.histplot()
# sns.displot()

sns.displot(data=tips, x="total_bill", kde=True)

plt.suptitle("Displot: Distribution of Total Bill", y=1.02)
plt.show()
```

```python
# ============================================================
# JOINTPLOT
# ============================================================

# Jointplot shows relationship between two numerical variables
# along with their individual distributions.

sns.jointplot(data=tips, x="total_bill", y="tip")

plt.show()
```

```python
# ============================================================
# JOINTPLOT WITH REGRESSION
# ============================================================

# kind='reg' adds a regression line.

sns.jointplot(data=tips, x="total_bill", y="tip", kind="reg")

plt.show()
```

```python
# ============================================================
# JOINTPLOT WITH HEX
# ============================================================

# kind='hex' shows density using hexagonal boxes.
# Darker boxes mean more data points.

sns.jointplot(data=tips, x="total_bill", y="tip", kind="hex")

plt.show()
```

```python
# ============================================================
# PAIRPLOT
# ============================================================

# Pairplot shows relationship between all numerical columns.
# It is very useful in EDA.

sns.pairplot(data=iris)

plt.show()
```

```python
# ============================================================
# PAIRPLOT WITH HUE
# ============================================================

# hue='species' shows different flower species in different colors.

sns.pairplot(data=iris, hue="species")

plt.show()
```

```python
# ============================================================
# RUGPLOT
# ============================================================

# Rugplot shows small lines for each observation on the axis.
# It helps us see where values are concentrated.

sns.rugplot(data=tips, x="total_bill")

plt.title("Rugplot: Total Bill")
plt.show()
```

```python
# ============================================================
# RUGPLOT WITH HISTPLOT
# ============================================================

# Combining histplot and rugplot gives better understanding.

sns.histplot(data=tips, x="total_bill", kde=True)
sns.rugplot(data=tips, x="total_bill")

plt.title("Histplot with Rugplot")
plt.show()
```

```python
# ============================================================
# REGRESSION PLOTS
# ============================================================

# Regression plots are used to show relationship and trend
# between two numerical variables.

# Regression line helps us understand:
# If x increases, does y increase or decrease?
# Is the relationship positive or negative?

# Common regression functions:
# sns.regplot()
# sns.lmplot()
```

```python
# ============================================================
# SIMPLE LINEAR REGRESSION PLOT
# ============================================================

# regplot creates scatter plot with regression line.

sns.regplot(data=tips, x="total_bill", y="tip")

plt.title("Simple Linear Regression Plot")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()

# Observation:
# The regression line is upward.
# This means tip generally increases as total bill increases.
```

```python
# ============================================================
# REGRESSION PLOT WITH HUE
# ============================================================

# regplot does not directly support hue.
# lmplot supports hue.

sns.lmplot(data=tips, x="total_bill", y="tip", hue="sex")

plt.title("Regression Plot with Hue")
plt.show()
```

```python
# ============================================================
# REGRESSION PLOT WITH HUE AND MARKERS
# ============================================================

# markers are used to change the shape of points.

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    markers=["o", "x"]
)

plt.title("Regression Plot with Hue and Markers")
plt.show()
```

```python
# ============================================================
# SETTING SIZE AND COLOR OF REGRESSION PLOT
# ============================================================

# scatter_kws controls scatter point properties.
# line_kws controls regression line properties.

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter_kws={"s": 80, "alpha": 0.7},
    line_kws={"linewidth": 3}
)

plt.title("Regression Plot with Custom Size")
plt.show()
```

```python
# ============================================================
# DISPLAYING MULTIPLE REGRESSION PLOTS
# ============================================================

# col='time' creates separate plots for Lunch and Dinner.
# This helps us compare two categories.

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="time"
)

plt.show()
```

```python
# ============================================================
# MULTIPLE REGRESSION PLOTS WITH HUE
# ============================================================

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time"
)

plt.show()
```

```python
# ============================================================
# SIZE AND ASPECT RATIO OF REGRESSION PLOT
# ============================================================

# height controls height of plot.
# aspect controls width compared to height.

# Formula:
# width = height * aspect

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    height=5,
    aspect=1.5
)

plt.show()
```

```python
# ============================================================
# MATRIX PLOTS
# ============================================================

# Matrix plots show data in rows and columns using color.
# They are useful for correlation and pattern detection.

# Common matrix plots:
# 1. Heatmap
# 2. Cluster Map
```

```python
# ============================================================
# CREATING CORRELATION MATRIX
# ============================================================

# Correlation shows relationship between numerical columns.
# Value ranges from -1 to +1.

# +1 means strong positive relationship.
# -1 means strong negative relationship.
#  0 means no relationship.

numeric_tips = tips.select_dtypes(include=["float64", "int64"])

correlation = numeric_tips.corr()

correlation
```

```python
# ============================================================
# HEATMAP
# ============================================================

# Heatmap uses colors to represent values.
# annot=True writes values inside boxes.
# cmap changes color theme.
# linewidths gives gap between cells.

sns.heatmap(correlation, annot=True, cmap="coolwarm", linewidths=0.5)

plt.title("Heatmap: Correlation Matrix")
plt.show()
```

```python
# ============================================================
# HEATMAP WITHOUT ANNOTATION
# ============================================================

sns.heatmap(correlation, cmap="YlGnBu")

plt.title("Heatmap Without Annotation")
plt.show()
```

```python
# ============================================================
# CLUSTER MAP
# ============================================================

# Cluster map groups similar rows and columns together.
# It is useful when we want to find similar patterns.

sns.clustermap(correlation, annot=True, cmap="coolwarm")

plt.show()
```

```python
# ============================================================
# STYLE AND COLOR IN SEABORN
# ============================================================

# Seaborn allows us to change background style of plots.

# Common styles:
# 1. white
# 2. ticks
# 3. darkgrid
# 4. whitegrid

# Syntax:
# sns.set_style("style_name")
```

```python
# ============================================================
# SET BACKGROUND TO WHITE
# ============================================================

sns.set_style("white")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Style: White Background")
plt.show()
```

```python
# ============================================================
# SET BACKGROUND TO TICKS
# ============================================================

sns.set_style("ticks")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Style: Ticks")
plt.show()
```

```python
# ============================================================
# SET BACKGROUND TO DARKGRID
# ============================================================

sns.set_style("darkgrid")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Style: Darkgrid")
plt.show()
```

```python
# ============================================================
# SET BACKGROUND TO WHITEGRID
# ============================================================

sns.set_style("whitegrid")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Style: Whitegrid")
plt.show()
```

```python
# ============================================================
# REMOVE AXES SPINE USING DESPINE
# ============================================================

# Axes spines are borders around the plot.
# sns.despine() removes top and right border by default.
# It makes chart look cleaner.
```

```python
# ============================================================
# DESPINE EXAMPLE
# ============================================================

sns.set_style("ticks")

sns.boxplot(data=tips, x="day", y="total_bill")

sns.despine()

plt.title("Boxplot with Despine")
plt.show()
```

```python
# ============================================================
# DESPINE WITH LEFT AND BOTTOM
# ============================================================

sns.boxplot(data=tips, x="day", y="total_bill")

sns.despine(left=True, bottom=True)

plt.title("Despine: Left and Bottom Removed")
plt.show()
```

```python
# ============================================================
# SIZE AND ASPECT
# ============================================================

# Plot size can be controlled in two ways:

# 1. Non-grid plots:
# Use plt.figure(figsize=(width, height))

# Examples:
# scatterplot, barplot, boxplot, heatmap

# 2. Grid type plots:
# Use height and aspect

# Examples:
# lmplot, relplot, catplot, displot
```

```python
# ============================================================
# NON-GRID PLOT SIZE
# ============================================================

# For normal plots, use plt.figure(figsize=(width, height)).

plt.figure(figsize=(10, 6))

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")

plt.title("Non-Grid Plot with Custom Figure Size")
plt.show()
```

```python
# ============================================================
# GRID TYPE PLOT SIZE
# ============================================================

# For figure-level plots like lmplot, use height and aspect.

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    height=5,
    aspect=2
)

plt.title("Grid Type Plot with Height and Aspect")
plt.show()
```

```python
# ============================================================
# SCALE AND CONTEXT
# ============================================================

# Seaborn context controls the size of plot elements.

# Common contexts:
# 1. paper
# 2. notebook
# 3. talk
# 4. poster

# Syntax:
# sns.set_context("context_name")

# paper    : Small size, useful for research papers.
# notebook : Default size, useful for Jupyter Notebook.
# talk     : Larger size, useful for presentation.
# poster   : Very large size, useful for posters.
```

```python
# ============================================================
# CONTEXT: PAPER
# ============================================================

sns.set_context("paper")
sns.set_style("whitegrid")

sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Context: Paper")
plt.show()
```

```python
# ============================================================
# CONTEXT: NOTEBOOK
# ============================================================

sns.set_context("notebook")
sns.set_style("whitegrid")

sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Context: Notebook")
plt.show()
```

```python
# ============================================================
# CONTEXT: TALK
# ============================================================

sns.set_context("talk")
sns.set_style("whitegrid")

sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Context: Talk")
plt.show()
```

```python
# ============================================================
# CONTEXT: POSTER
# ============================================================

sns.set_context("poster")
sns.set_style("whitegrid")

sns.barplot(data=tips, x="day", y="total_bill")

plt.title("Context: Poster")
plt.show()
```

```python
# ============================================================
# FIGURE-LEVEL FUNCTIONS IN SEABORN
# ============================================================

# Seaborn has two types of functions:

# 1. Axes-Level Functions
# These create one plot on matplotlib axes.

# Examples:
# sns.scatterplot()
# sns.lineplot()
# sns.barplot()
# sns.boxplot()
# sns.heatmap()

# 2. Figure-Level Functions
# These can create multiple plots using FacetGrid.

# Examples:
# sns.relplot()
# sns.catplot()
# sns.displot()
# sns.lmplot()
```

```python
# ============================================================
# RELPLOT
# ============================================================

# relplot is figure-level function for relational plots.
# kind='scatter' creates scatter plot.
# col creates multiple plots.

sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time",
    kind="scatter"
)

plt.show()
```

```python
# ============================================================
# CATPLOT
# ============================================================

# catplot is figure-level function for categorical plots.
# kind='box' creates boxplot.
# col='time' creates separate plots.

sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    kind="box"
)

plt.show()
```

```python
# ============================================================
# DISPLOT WITH MULTIPLE PLOTS
# ============================================================

# displot can create multiple distribution plots using col.

sns.displot(
    data=tips,
    x="total_bill",
    col="time",
    kde=True
)

plt.show()
```

```python
# ============================================================
# LMPLOT WITH MULTIPLE PLOTS
# ============================================================

sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time"
)

plt.show()
```

| Plot Category           | Plot Name    | Function            | Main Use                                        |
| ----------------------- | ------------ | ------------------- | ----------------------------------------------- |
| Relational              | Scatter Plot | `sns.scatterplot()` | Relationship between two numerical variables    |
| Relational              | Line Plot    | `sns.lineplot()`    | Trend over time                                 |
| Relational / Regression | lmplot       | `sns.lmplot()`      | Scatter plot with regression line               |
| Categorical             | Stripplot    | `sns.stripplot()`   | Individual data points by category              |
| Categorical             | Swarmplot    | `sns.swarmplot()`   | Non-overlapping individual points               |
| Categorical             | Barplot      | `sns.barplot()`     | Average numerical value by category             |
| Categorical             | Countplot    | `sns.countplot()`   | Count/frequency of category                     |
| Categorical             | Boxplot      | `sns.boxplot()`     | Distribution, median and outliers               |
| Categorical             | Violinplot   | `sns.violinplot()`  | Distribution shape by category                  |
| Distribution            | Histplot     | `sns.histplot()`    | Frequency distribution                          |
| Distribution            | Displot      | `sns.displot()`     | Figure-level distribution plot                  |
| Distribution            | Jointplot    | `sns.jointplot()`   | Relationship plus distribution                  |
| Distribution            | Pairplot     | `sns.pairplot()`    | Pairwise relationship between numerical columns |
| Distribution            | Rugplot      | `sns.rugplot()`     | Shows individual values on axis                 |
| Regression              | Regplot      | `sns.regplot()`     | Regression line with scatter plot               |
| Matrix                  | Heatmap      | `sns.heatmap()`     | Matrix/correlation visualization                |
| Matrix                  | Cluster Map  | `sns.clustermap()`  | Grouped heatmap                                 |
